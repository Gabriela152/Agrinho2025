```python
banco_dados = []

def menu():
    print("\n=== MENU PRINCIPAL ===")
    print("1 - Adicionar registro")
    print("2 - Listar registros")
    print("3 - Buscar por nome")
    print("4 - Remover registro")
    print("5 - Sair")

while True:
    menu()
    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        nome = input("Digite o nome: ")
        idade = int(input("Digite a idade: "))
        cidade = input("Digite a cidade: ")

        # Cada pessoa será um dicionário
        pessoa = {"nome": nome, "idade": idade, "cidade": cidade}
        banco_dados.append(pessoa)

        print("Registro adicionado com sucesso!")

    elif opcao == "2":
        print("\n=== LISTA DE REGISTROS ===")
        if not banco_dados:
            print("Nenhum registro encontrado.")
        else:
            for i, pessoa in enumerate(banco_dados, start=1):
                print(f"{i}. Nome: {pessoa['nome']}, Idade: {pessoa['idade']}, Cidade: {pessoa['cidade']}")

    elif opcao == "3":
        nome_busca = input("Digite o nome para buscar: ")
        encontrado = False
        for pessoa in banco_dados:
            if pessoa["nome"].lower() == nome_busca.lower():
                print(f"Registro encontrado: {pessoa}")
                encontrado = True
        if not encontrado:
            print("Nenhum registro com esse nome foi encontrado.")

    elif opcao == "4":
        nome_remove = input("Digite o nome para remover: ")
        for pessoa in banco_dados:
            if pessoa["nome"].lower() == nome_remove.lower():
                banco_dados.remove(pessoa)
                print("🗑️ Registro removido com sucesso!")
                break
        else:
            print("Nome não encontrado.")

    elif opcao == "5":
        print("Encerrando o programa...")
        break

    else:
        print("Opção inválida! Tente novamente.")

# Programa: Banco de Dados Simples em Python

# Lista que servirá como "banco de dados"
banco_dados = []

def menu():
    print("\n=== MENU PRINCIPAL ===")
    print("1 - Adicionar registro")
    print("2 - Listar registros")
    print("3 - Buscar por nome")
    print("4 - Remover registro")
    print("5 - Sair")

while True:
    menu()
    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        nome = input("Digite o nome: ")
        idade = int(input("Digite a idade: "))
        cidade = input("Digite a cidade: ")

        # Cada pessoa será um dicionário
        pessoa = {"nome": nome, "idade": idade, "cidade": cidade}
        banco_dados.append(pessoa)

        print("Registro adicionado com sucesso!")

    elif opcao == "2":
        print("\n=== LISTA DE REGISTROS ===")
        if not banco_dados:
            print("Nenhum registro encontrado.")
        else:
            for i, pessoa in enumerate(banco_dados, start=1):
                print(f"{i}. Nome: {pessoa['nome']}, Idade: {pessoa['idade']}, Cidade: {pessoa['cidade']}")

    elif opcao == "3":
        nome_busca = input("Digite o nome para buscar: ")
        encontrado = False
        for pessoa in banco_dados:
            if pessoa["nome"].lower() == nome_busca.lower():
                print(f"✅ Registro encontrado: {pessoa}")
                encontrado = True
        if not encontrado:
            print("Nenhum registro com esse nome foi encontrado.")

    elif opcao == "4":
        nome_remove = input("Digite o nome para remover: ")
        for pessoa in banco_dados:
            if pessoa["nome"].lower() == nome_remove.lower():
                banco_dados.remove(pessoa)
                print("🗑️ Registro removido com sucesso!")
                break
        else:
            print(" Nome não encontrado.")

    elif opcao == "5":
        print("Encerrando o programa...")
        break

    else:
        print(" Opção inválida! Tente novamente.")
```
